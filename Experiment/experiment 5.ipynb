```python
import numpy as np
import pandas as pd
from collections import Counter

# Dataset: Iris - UCI Machine Learning Repository (iris.csv)
# ---- 1. Load CSV from online URL to ensure it runs ----
url = 'https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv'
df = pd.read_csv(url)

# Map columns appropriately (gist iris.csv columns: sepal.length, sepal.width, petal.length, petal.width, variety)
X = df.iloc[:, :-1].values.astype(float)
y = df.iloc[:, -1].values

# ---- 2. Normalize inputs (0 to 1) ----
X = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))

# ---- 3. Train / test split (80% / 20%) ----
np.random.seed(42)
idx = np.random.permutation(len(X))
split = int(0.8 * len(X))
Xtr, Xte = X[idx[:split]], X[idx[split:]]
ytr, yte = y[idx[:split]], y[idx[split:]]
print("Train size:", len(Xtr), " Test size:", len(Xte), "\n")

# ---- 4. K-NN algorithm ----
def euclidean(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def knn_predict(x, k):
    # distance from x to every training point
    dists = [euclidean(x, t) for t in Xtr]
    # indices of the k smallest distances
    nearest = np.argsort(dists)[:k]
    # majority vote among the k neighbours
    votes = Counter(ytr[nearest])
    return votes.most_common(1)[0][0]

# ---- 5. Test with different values of K ----
print("K   Accuracy")
for k in [1, 3, 5, 7, 9]:
    preds = [knn_predict(x, k) for x in Xte]
    acc = np.mean(np.array(preds) == yte) * 100
    print(f"{k}   {acc:.2f}%")

# ---- 6. Detailed result for K = 5 ----
k = 5
preds = [knn_predict(x, k) for x in Xte]
print(f"\nSample predictions (K = {k}):")
for i in range(8):
    mark = "correct" if preds[i] == yte[i] else "WRONG"
    print(f"  Actual: {yte[i]:<12} Predicted: {preds[i]:<12} {mark}")

# ---- 7. Classify a new sample ----
new_sample = np.array([5.9, 3.0, 5.1, 1.8])          # raw values
raw = df.iloc[:, :-1].values.astype(float)
new_norm = (new_sample - raw.min(axis=0)) / (raw.max(axis=0) - raw.min(axis=0))
print("\nNew sample:", new_sample.tolist())
print("Predicted class:", knn_predict(new_norm, k))
```
